```python
# Create array with zeros
z = np.zeros((3,3))

# Create array with ones
o = np.ones((3,3))

# Create array with random values
r = np.random.rand(3,3) # only 0-1

#Create array with random values with range
r2 = np.random.randint(1,10, size=(3,3))

#Create identical matrix
i = np.eye(3)


print(f"Zeros:\n {z}")
print(f"Ones:\n {o}")
print(f"Random:\n {r}")
print(f"Random with range:\n {r2}")
print(f"Identical:\n {i}")
```

## Copy & View

```python
arr=np.array([1,2,3,4,5])
print(arr)
```

```python
# Copy
new_arr=arr.copy()
print(new_arr)
```

```python
arr[0]=9
print(arr)
print(new_arr) # arr doesn't change when we use copy()
```

```python
arr=np.array([1,2,3,4,5])
new_arr=arr.view()
print(new_arr)
print(arr)
```

```python
arr[0]=9
print(new_arr)
print(arr) #  arr change when we use copy()
```

## Join & Split

```python
arr1=np.array([1,2,3,4])
arr2=np.array([4,5,6,7,8])
```

```python
new_arr=np.concatenate((arr1,arr2))
print(new_arr)
```

```python
arr1=np.array([[1,2],[3,4]])
arr2=np.array([[5,6],[7,8]])

new_arr=np.concatenate((arr1,arr2),axis=1)
print(new_arr)
new_arr=np.concatenate((arr1,arr2))
print(f"\n{new_arr}")
```

```python
# Split
arr=np.array([1,2,3,4,5,6])
new_arr=np.array_split(arr,2)
print(new_arr)
print(new_arr[0])
print(new_arr[1])
```

```python
# Datafames join

df1 = pd.DataFrame({
    "Name": ["kamal", "saman", "nimal"],
    "Age": [20, 32, 26],
    "Mark": [85, 67, 73]
})

df1.set_index("Name", inplace=True)

print("df1:")
print(df1)
```

```python

df2 = pd.DataFrame({
    "Name": ["kamal", "saman", "pawan"],
    "District": ["Colombo", "Kandy", "Galle"],
    "Mark": [85, 67, 50]
})

df2.set_index("Name", inplace=True)

print("df2:")
print(df2)
```

```python
df1.join(df2, lsuffix='_df1', rsuffix='_df2', how='left')
```

```python
df1.join(df2, lsuffix='_df1', rsuffix='_df2', how='right')
```

```python
df1.join(df2, lsuffix='_df1', rsuffix='_df2', how='inner')
```

```python
df1.join(df2, lsuffix='_df1', rsuffix='_df2', how='outer')
```

```python
# Concatenate
pd.concat([df1,df2], join='inner')
```

```python
pd.concat([df1,df2], join='inner', axis=1)
```

```python
pd.concat([df1,df2], join='outer')
```

```python
pd.concat([df1,df2], join='outer',axis=1)
```

## Sub plots

```python
x=np.arange(1,10,0.1)
y1=x
y2=np.sqrt(x)
y3=np.power(x,2)
y4=np.power(x,3)
```

```python
plt.figure()
plt.subplot(2,2,1)
plt.plot(x,y1,'k--')

plt.subplot(2,2,2)
plt.plot(x,y2,'r--')

plt.subplot(2,2,3)
plt.plot(x,y3,'m:')

plt.subplot(2,2,4)
plt.plot(x,y4)

plt.show()
```
